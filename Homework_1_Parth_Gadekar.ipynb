## Question 1
Print the result of **A+B** and explain the mechanism behind this.

You answer:

```python
A = np.array([[0, 1, 2]])
B = np.array([[10, 10, 10], [20, 20, 20], [30, 30, 30]])
# insert your code and provide a description of the underlying mechanism called broadcast

result = A + B
result
```

Broadcasting is a powerful mechanism in numpy that allows operations on arrays of different shapes.
Numpy compares the shapes element-wise starting from the trailing dimensions.
In case of this example, for the last dimension, both the arrays have size 3, which are compatible.
But for the first dimension, A has size 1 while B has size 3.
Numpy automatically stretches or you can say broadcasts A along the first dimension so that it behaves as if it is an array of shape (3, 3) where each row is identical to the original row in A. After broadcasting, the addition is performed element-wise.

##Question 2
Plot MSE versus order m, for m = 1, 2, 3, 4, 5, 6, 7, 8, 9 respectively. Identify the best choice of m.
Discuss how m impacts the performance in terms model accuracy and overfitting

```python
# Initialize lists for storing results
m_values = range(1, 10)  # Polynomial orders from 1 to 9
mse_values = []

# Fit polynomials and compute MSE for each order
for m in m_values:
    coefficients = np.polyfit(x, y, m)
    polynomial = np.poly1d(coefficients)
    y_pred = polynomial(x)
    mse = np.mean((y - y_pred) ** 2)
    mse_values.append(mse)

# Identify the best polynomial order (minimum MSE)
best_m = m_values[np.argmin(mse_values)]

# Plot MSE vs. polynomial order
plt.figure(figsize=(8, 5))
plt.plot(m_values, mse_values, marker='o', color='b', label='MSE')
plt.axvline(best_m, color='r', linestyle='--', label=f'Best m = {best_m}')
plt.title('MSE vs. Polynomial Order (m)')
plt.xlabel('Polynomial Order (m)')
plt.ylabel('Mean Squared Error (MSE)')
plt.legend()
plt.grid(True)
plt.show()

# Print the best m
print(f"The best polynomial order (m) is {best_m}, with a minimum MSE of {min(mse_values):.2f}.")
```

The best polynomial order m is the one that minimizes the Mean Squared Error (MSE), as identified in the plot of MSE vs. Polynomial Order. From the calculations, the optimal value of m balances accuracy without overfitting. The choice of m directly influences the model's ability to generalize. It must be chosen carefully to achieve an appropriate balance between bias and variance.
